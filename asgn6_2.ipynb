```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temp_adjusted = temperature + 2
print("a. Adjusted temperatures (+2°C):", temp_adjusted)

fahrenheit = (9 / 5) * temperature + 32
print("b. Temperatures in Fahrenheit:", fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("c. Readings > 32°C:", greater_than_32)

count_exceed_32 = np.sum(temperature > 32)
print("d. Count of readings > 32°C:", count_exceed_32)

print(
    "\ne. Vectorization executes operations at the C-level in compiled code, eliminating Python loop overhead and dynamic type checking. Boolean indexing evaluates conditions across the entire array simultaneously to generate a mask, avoiding element-by-element conditional checks."
)
```

```python
import numpy as np

steps = np.array(
    [
        [5000, 6200, 7100],
        [8000, 7500, 9000],
        [4500, 5100, 4800],
        [9000, 8500, 9500],
    ]
)

total_steps = np.sum(steps)
print("a. Total steps recorded:", total_steps)

mean_steps = np.mean(steps)
print("b. Mean number of steps:", mean_steps)

max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"c. Maximum steps: {max_steps} | Minimum steps: {min_steps}")

total_steps_per_day = np.sum(steps, axis=0)
print("d. Total steps for each day (axis=0):", total_steps_per_day)

total_steps_per_user = np.sum(steps, axis=1)
print("e. Total steps for each user (axis=1):", total_steps_per_user)

max_pos_flat = np.argmax(steps)
max_pos_2d = np.unravel_index(max_pos_flat, steps.shape)
print("f. Position of max steps (flat index):", max_pos_flat)
print("f. Position of max steps (user, day):", max_pos_2d)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original array:", original)

subset = original[1:5]
print("b. Subset (index 1 to 4):", subset)

subset[0] = 999
print("c. Subset:", subset)
print("   Original:", original)

copied_array = original[1:5].copy()
copied_array[0] = 500
print("d. Copied array:", copied_array)
print("   Original:", original)

matrix_3x4 = np.arange(1, 13).reshape(3, 4)
print("e. 3x4 Matrix:\n", matrix_3x4)

print("f. First row:", matrix_3x4[0])
print("   Last row:", matrix_3x4[-1])
print("   Second column:", matrix_3x4[:, 1])
print("   Elements from rows 1-2 and columns 2-3:\n", matrix_3x4[1:3, 2:4])

print("g. flatten():", matrix_3x4.flatten())
print("   ravel():", matrix_3x4.ravel())

matrix_3x4_h = np.arange(1, 13).reshape(3, 4)
rav_arr = matrix_3x4_h.ravel()
rav_arr[0] = 999
print("h. Ravel modification - Original affected?", matrix_3x4_h[0, 0] == 999)

matrix_3x4_i = np.arange(1, 13).reshape(3, 4)
flat_arr = matrix_3x4_i.flatten()
flat_arr[0] = 999
print("i. Flatten modification - Original affected?", matrix_3x4_i[0, 0] == 999)

print("j. Shape:", matrix_3x4.shape)
print("   ndim:", matrix_3x4.ndim)
print("   Size:", matrix_3x4.size)
print("   dtype:", matrix_3x4.dtype)
```

```python
import numpy as np

X = np.array([[6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8]])
y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape)
print("   Dimensions (ndim) of X:", X.ndim)

X_T = X.T
print("b. Transpose of X (X.T):\n", X_T)
print(
    "   Explanation: The transpose swaps rows and columns so rows represent the three features (Sleep, Activity, Stress) and columns represent individual users."
)

XT_X = X.T @ X
print("c. Matrix product X.T @ X:\n", XT_X)

XT_X_inv = np.linalg.inv(XT_X)
print("d. Inverse of (X.T @ X):\n", XT_X_inv)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("e. OLS Coefficients (beta):", beta)

print(
    "f. Explanation: Each coefficient represents the expected change in the user's assistance score for a one-unit increase in that respective feature, holding all other features constant."
)

new_user = np.array([5, 40, 7])
predicted_score = np.dot(new_user, beta)
print("g. Predicted assistance score for new user:", predicted_score)
```
