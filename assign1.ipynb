## 6.2 Prime number

```python
def IsPrime(n):
    if n < 2:
        return 0
    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return 0
    return 1

print("IsPrime(20)  --> ", IsPrime(20))
print("IsPrime(23)  --> ", IsPrime(23))
print("IsPrime(200) --> ", IsPrime(200))
print("IsPrime(37)  --> ", IsPrime(37))
```

### Assignment 6.2

**Task:** WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
def IsPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def AddPrimes(n):
    s = 0
    for i in range(2, n + 1):
        if IsPrime(i):
            s += i
    return s

n = int(input("Enter n: "))
print("Sum of prime numbers is -->", AddPrimes(n))
```
