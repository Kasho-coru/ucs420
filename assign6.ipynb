```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid")
```

```python
#q1
M = float(input("Enter a value for M: "))

x = np.linspace(-10, 10, 100)

y1 = M*(x ** 2)
y2 = M*np.sin(x)
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label=f'y = {M} x^2', color='crimson', linestyle='--', linewidth=2)
plt.plot(x, y2, label=f'y = {M} sin(x)', color='navy', linestyle='-', linewidth=2)
plt.title(f'Comparison of y = M  x^2 and y = M sin(x) for M = {M}', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)

plt.show()
```

```python
#q2
data = {
    'Subject': ['Mathematics', 'Physics', 'Chemistry', 'Computer Science', 'English'],
    'Score': [88, 75, 92, 95, 81]
}
df_scores = pd.DataFrame(data)
plt.figure(figsize=(9, 5))
ax = sns.barplot(x='Subject', y='Score', data=df_scores, palette='Set2', hue='Subject', legend=False)
plt.title('Subject Scores Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Subjects', fontsize=12)
plt.ylabel('Scores', fontsize=12)
plt.ylim(0, 110)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
#q3
roll_number = 1024170068
np.random.seed(roll_number)
data_q3 = np.random.randn(50)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes[0, 0].plot(np.cumsum(data_q3), color='teal', marker='o', linestyle='-')
axes[0, 0].set_title('Cumulative Sum')
axes[0, 0].set_xlabel('Index')
axes[0, 0].set_ylabel('Cumulative Value')
axes[0, 0].grid(True, linestyle='--')
noise = np.random.randn(50) * 0.5
axes[0, 1].scatter(range(50), data_q3 + noise, color='coral', alpha=0.8)
axes[0, 1].set_title('Scatter Plot with Random Noise')
axes[0, 1].set_xlabel('Index')
axes[0, 1].set_ylabel('Noise')
axes[0, 1].grid(True, linestyle='--')
axes[1, 0].hist(data_q3, bins=10, color='mediumpurple', edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Histogram')
axes[1, 0].set_xlabel('Range')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].grid(True, linestyle='--')
sns.boxplot(y=data_q3, ax=axes[1, 1], color='skyblue')
axes[1, 1].set_title('Box Plot')
axes[1, 1].set_ylabel('Value')
axes[1, 1].grid(True, linestyle='--')

plt.tight_layout()
plt.show()
```

```python
#q4
from google.colab import files
uploaded=files.upload()
df=pd.read_csv(next(iter(uploaded)))
plt.figure(figsize=(10, 4))
sns.lineplot(x='month_number', y='total_profit', data=df, marker='o', color='darkgreen', linewidth=2.5)
plt.title('Total Profit of All Months', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Profit', fontsize=12)
plt.xticks(df['month_number'])
plt.grid(True, linestyle='--')
plt.show()
products = ['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer']

plt.figure(figsize=(12, 6))
for product in products:
    sns.lineplot(x='month_number', y=product, data=df, label=product, marker='o')

plt.title('Monthly Sales Data for All Products', fontsize=14)
plt.xlabel('Month Number', fontsize=12)
plt.ylabel('Units Sold', fontsize=12)
plt.xticks(df['month_number'])
plt.grid(True, linestyle='--')
plt.legend(title='Products')
plt.show()
df_melted = df.melt(id_vars=['month_number'], var_name='Attribute', value_name='Value')

plt.figure(figsize=(14, 7))
sns.barplot(x='month_number', y='Value', hue='Attribute', data=df_melted, palette='tab10')
plt.title('Monthly Comparison of All Attributes', fontsize=14)
plt.xlabel('Month Number', fontsize=12)
plt.ylabel('Value / Units', fontsize=12)
plt.legend(title='Attributes', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--')
plt.tight_layout()
plt.show()
```
