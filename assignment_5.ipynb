```python
import numpy as np
```

```python
#q1
arr1 = np.array([10, 20, 30, 40, 50])

#a)
arr1_add = arr1+2

#b)
arr1_mul = arr1*3

#c)
arr1_div = arr1/2

print("Original Array:", arr1)
print("a) + 2:", arr1_add)
print("b) * 3:", arr1_mul)
print("c) / 2:", arr1_div)
```

```python
#q2
#a)
arr2_a = np.array([1, 2, 3, 6, 4, 5])
reversed_arr2_a = arr2_a[::-1]
print("Reversed Array:", reversed_arr2_a)

#b)
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print("Most frequent value:", most_freq_x)
print("Indices:", indices_x)
#c)
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_freq_y = vals_y[counts_y == max_count_y]

print("Most frequent values:", most_freq_y)
for val in most_freq_y:
    print(f"Indices for value {val}:", np.where(y == val)[0])
```

```python
#q3
arr3 = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

#a)
elem_a = arr3[0, 1]

#b)
elem_b = arr3[2, 0]

print("Array:\n", arr3)
print("a) 1st row, 2nd column:", elem_a)
print("b) 3rd row, 1st column:", elem_b)
```

```python
#q4
Coral = np.linspace(10, 100, 25)

print("Array:", Coral)
print("Dimensions:", Coral.ndim)
print("Shape:", Coral.shape)
print("Total Elements:", Coral.size)
print("Data Type:", Coral.dtype)
print("Total Bytes Consumed:", Coral.nbytes)

transposed_via_reshape = Coral.reshape(25, 1)
print("\nTransposed via reshape (25x1 shape):\n", transposed_via_reshape)

```

```python
#q5
ucs420_Coral = np.array(
    [[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])

print("Original Array (3x4):\n", ucs420_Coral)

print("Mean:", np.mean(ucs420_Coral))
print("Median:", np.median(ucs420_Coral))
print("Max:", np.max(ucs420_Coral))
print("Min:", np.min(ucs420_Coral))
print("Unique Elements:", np.unique(ucs420_Coral))

reshaped_ucs420_Coral = ucs420_Coral.reshape(4, 3)
print("\nReshaped Array (4x3):\n", reshaped_ucs420_Coral)

resized_ucs420_Coral = np.resize(ucs420_Coral, (2, 3))
print("\nResized Array (2x3):\n", resized_ucs420_Coral)
```
