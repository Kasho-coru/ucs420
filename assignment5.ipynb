```python
#q1
import numpy as np
import matplotlib.pyplot as plt
gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')
sum=np.sum(gfg)
sum_row = np.sum(gfg, axis=1)
sum_col = np.sum(gfg, axis=0)
print(f"i. Sum of all elements: {sum}")
print(f"ii. Sum row-wise: {sum_row}")
print(f"iii. Sum column-wise: {sum_col}")
```

```python
#q2a
arr_a = np.array([10, 52, 62, 16, 16, 54, 453])
sorted_arr = np.sort(arr_a)
sorted_indices = np.argsort(arr_a)
smallest_4 = sorted_arr[:4]
largest_5 = sorted_arr[-5:]
print(f"i. Sorted array: {sorted_arr}")
print(f"ii. Indices of sorted array: {sorted_indices}")
print(f"iii. 4 smallest elements: {smallest_4}")
print(f"iv. 5 largest elements: {largest_5}")
```

```python
#q2b
arr_b = np.array([1.0, 1.2, 2.2, 2.0, 3.0, 2.0])
integers_only = arr_b[arr_b == arr_b.astype(int)]
floats_only = arr_b[arr_b != arr_b.astype(int)]
print(f"i. Integers only: {integers_only}")
print(f"ii. Floats only: {floats_only}")
```

```python
#q3
first_name = "Coral"
last_name = "Singh"
X = ord(first_name[0].upper()) + ord(last_name[0].upper())
sales = np.array([X, X + 50, X + 100, X + 150, X + 200])
tax_rate = ((X % 5) + 5) / 100
sales_after_tax = sales * (1 + tax_rate)
discounts = np.where(sales < (X + 100), 0.05, 0.10)
discounted_sales = sales * (1 - discounts)
weeks_matrix = np.tile(sales, (3, 1))
week_multipliers = np.array([[1.00], [1.02], [1.04]])
weekly_increased_sales = weeks_matrix * week_multipliers
print(f"a) X value (ASCII sum): {X}")
print(f"   Base Sales Array: {sales}")
print(f"b) Tax Rate: {tax_rate * 100:.1f}%")
print(f"   Sales after Tax: {sales_after_tax}")
print(f"c) Sales after Discount: {discounted_sales}")
print("\nd) 3-Week Sales Matrix with 2% weekly increase:")
print(weekly_increased_sales)
```

```python
#q4
x = np.linspace(-10, 10, 100)

y_sq = x**2
y_sin = np.sin(x)
y_exp = np.exp(x)
y_log = np.log(np.abs(x) + 1)

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.plot(x, y_sq, color='blue')
plt.title(r'$Y = x^2$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.subplot(2, 2, 2)
plt.plot(x, y_sin, color='green')
plt.title(r'$Y = \sin(x)$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.subplot(2, 2, 3)
plt.plot(x, y_exp, color='red')
plt.title(r'$Y = e^x$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.subplot(2, 2, 4)
plt.plot(x, y_log, color='purple')
plt.title(r'$Y = \log(|x| + 1)$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)

plt.tight_layout()
plt.show()
```
